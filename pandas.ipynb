```python
import pandas as pd

data = {
    'Name': ['Akshay', 'Rahul', 'Deepu', 'Arjun'],
    'Age': [19, 20, 19, 21],
    'Marks': [85, 92, 76, 88]
}

df = pd.DataFrame(data)

print(df[df['Marks'] > 80])
```

```python
import pandas as pd

data = {
    'Name': ['Akshay', 'Rahul', 'Deepu', 'Arjun'],
    'Age': [19, 20, 19, 21],
    'Marks': [85, 92, 76, 88]
}

df = pd.DataFrame(data)

print(df.sort_values(by='Marks', ascending=False))
```

```python
import pandas as pd

data = {
    "Name": ["Akshay", "Rahul", "Deepu", "Arjun"],
    "Marks": [85, None, 76, None]
}

df = pd.DataFrame(data)

print(df.isnull().sum())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```

```python
import pandas as pd

data = {
    "Department": ["CSE", "ECE", "CSE", "ECE", "CSE"],
    "Marks": [85, 78, 92, 88, 75]
}

df = pd.DataFrame(data)
print(df.groupby("Department")["Marks"].mean())
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Akshay", "Rahul", "Deepu"]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    "Marks": [85, 92, 76]
}, index=[1, 2, 3])

result = df1.join(df2)

print(result)
```
