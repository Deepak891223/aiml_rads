Creating a numpy array

```python
import numpy as np
a = np.array([10,20,30,40,50])
print(a)
```

indexing of array

```python
import numpy as np
a = np.array([10,20,30,40,50])
print(a[0])
print(a[1])
print(a[2])
print(a[3])
print(a[-2])
print(a[-1])
```

slicing of an array

```python
import numpy as np
a = np.array([10,20,30,40,50])
print(a[0:])
print(a[2:])
print(a[3:4])
```

shape

```python
import numpy as np
a = np.array([[1,2,3],[4,5,6]])
print(a.shape)
```

Dimensions

```python
import numpy as np
a=np.array([[1,2,3],[4,5,6]])
print(a.ndim)
```

Reshaping

```python
import numpy as np
a = np.array([1,2,3,4,5,6])
new_a = a.reshape(2,3)
print(new_a)
```

BroadCasting

```python
import numpy as np
a = np.array([1,2,3,4])
print(a+5)
```

Vectorization

```python
import numpy as np
a = np.array([1,2,3,4,5])
print(a**2)
```

Matrix Multiplication

```python
import numpy as np
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
C = np.dot(A,B)
print(C)
```
