```python
def ForwardDialEncoding(n,m,positions):
  current = 1
  total=0
  for target in positions:
    direct = abs(target-current)
    distance = min(direct,n-direct)
    total += distance
    current = target
  return total
n = int(input())
m = int(input())
positions = list(map(int,input().split()))
print(ForwardDialEncoding(n,m,positions))
```
